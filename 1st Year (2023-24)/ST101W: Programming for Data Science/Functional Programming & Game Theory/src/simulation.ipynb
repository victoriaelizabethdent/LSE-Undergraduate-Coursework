### Instructions

In the code block below, write code to _estimate_ the probability via _simulation_ for the following question:

> A fair die is rolled until a 6 occurs five times. Calculate the probability that this random experiment finishes after at least 12 rolls have been played, with a 6 occurring on the 1st and the 2nd rolls.

Please also run the code and use the `print()` function to print out the resulting estimated probability.

```python
import random

num_success = 0
num_trials = 10000
dice = [1, 2, 3, 4, 5, 6]

for _ in range(num_trials):
    num_sixes = 0
    experiment = random.choices(dice, k = 100)
    for roll in experiment[:12]:
        if roll == 6:
            num_sixes += 1
    if experiment[0] == 6 and experiment[1] == 6 and num_sixes < 5:
        num_success += 1

probability = num_success/num_trials
print(f'the probability that the experiment finishes after at least 12 rolls have been played, with a 6 occurring on the 1st and the 2nd rolls is {probability}')
```

In the code, we assume that if each experiment reaches 12 rolls and the number of sixes is less than 5, then the experiment will be successful assuming inifite rolls of the die.
