5. (5 marks) Which programmes were suspended and discontinued in 2018-2022 (if there were any)?

    **The following programmes were discontinued or suspended in 2018-2022:** 
    - **'BSc in Statistics with Finance'**
    - **'BSc in Business Mathematics and Statistics'**
    - **'BSc in Social Policy and Economics'**
    - **'BSc in Social Policy and Sociology'**
    - **'BSc in Social Policy with Government'**
    - **'BSc in Social Policy'** 

```python
lse_grouped= lse_students_dataframe.groupby(['programme']).sum()

suspended= []
for year in range(2019, 2023):
    prev= year -1
    suspended.extend(lse_grouped[(lse_grouped[str(prev)]>0) & (lse_grouped[str(year)]==0)].index)
```

6. (6 marks) Which programme had the largest growth? Please also state how you are measuring "growth".

    **We measured growth in terms of the minimum amount of students the programme had between 2018-2022, compared to the amount of student observed today, in 2022. We found on these conditions, that BSc in International Social and Public Policy had the largest growth, of 190 students. Before calculating this, the dataframe was cleaned of any programmes that were discontinued, any ones that were just suspended were kept hence why we did not reuse the code from the previous question. This is because any discontinued programmes would have automatically have 0 growth, due to how we measured growth.**

```python
#removing all the discontinued programmes as they all had no growth 
lse_grouped = lse_grouped[lse_grouped['2022'] != 0]

min = lse_grouped.loc[:, '2018':'2022'].min(axis=1)
growth = lse_grouped['2022']- min
growth.max()
growth.idxmax()
```
